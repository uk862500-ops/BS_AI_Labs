```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
print("setup completed")
```

```python
#task1: Create a line graph showing how temperature changes
over 5 days.
#Data
days = ["Mon", "Tue", "Wed", "Thu", "Fri"]
temp = [30,32,31,35,33]
#graph theme
sns.set_theme(style="whitegrid")
#figure size
plt.figure(figsize = (10,6))
#plotting of days and temp
plt.plot(days, temp,marker = "o",linewidth = 5,color = "Red")
#add title
plt.title("Temprature of 5 Days",color = "Silver")
#add labels to x
plt.xlabel("Days",color = "Green",fontsize = 50)
#add labels to y
plt.ylabel("temp",color = "Green",fontsize = 50)
#show grid
plt.grid(True, linestyle = '--', alpha = 0.7)
plt.show()
```

```python
#task2:Create a bar chart to compare marks of 4 students.
plt.figure(figsize=(6, 4))
students = ['Ali', 'Sara', 'Ahmed', 'Ayesha']
marks = [78, 85, 90, 72]
colors = ['#FF9999', '#66B3FF', '#99FF99', '#FFCC99']

bars = plt.bar(students, marks, color=colors, edgecolor='grey')
plt.title('Student Marks Comparison')
plt.xlabel('Students')
plt.ylabel('Marks')
plt.ylim(0, 110)

# Display data values clearly on top of each bar
for bar in bars:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2.0, yval + 2, str(yval), 
             ha='center', va='bottom', fontweight='bold')

plt.tight_layout()
plt.show()
```

```python
#task3: Generate a histogram of 500 random exam scores.
np.random.seed(42)
score = np.random.randint(0,100,500)
plt.figure(figsize= (8,4))
plt.hist(score,bins = 10, color = "skyblue",edgecolor = "black")
plt.title("histogram of 500 random exam score")
plt.xlabel("Exam score")
plt.ylabel("no of student")
plt.show()
```

```python
#task4: Show relationship between study hours and marks.
#Data
hours = [1,2,3,4,5,6,7]
Marks = [40,45,50,60,70,80,90]
plt.figure(figsize = (8,4))
plt.scatter(hours,Marks, color = "Red")
plt.title("Relationship Between Study Hours and Marks")
plt.show()
```

```python
#Task5: Create a smooth line plot using Seaborn
x = [1,2,3,4,5]
y = [10,20,25,30,40]

fig, (left_plot, right_plot) = plt.subplots(1, 2, figsize=(12, 4.5))
left_plot.plot(x,y, color = "orange")
left_plot.set_title("Matplotlib plot line")

sns.lineplot(x = x, y = y, ax=right_plot, color='purple')
right_plot.set_title("seaborn smoth line plot")
plt.show()
```

```python
#Task6: Create a heatmap of a 5×5 random matrix.
matrix_data = np.random.rand(5,5)
sns.heatmap(matrix_data,annot=True , cmap='viridis', cbar=True)
plt.title("Heatmap of a 5x5 Random Matrix")
plt.show()
```

```python
#task6: Task: Plot same data using both libraries.
x = [1,2,3,4,5]
y =  [2,4,6,8,10]
fig, axes = plt.subplots(1, 2, figsize=(12, 4.5))
axes[0].scatter(x,y, color = "Red")
axes[0].set_title("Matplotlib scatter",color = "skyblue")

sns.scatterplot(x = x, y = y,ax = axes[1])
axes[1].set_title("seaborn scatters", color ="skyblue")
plt.show()
```

```python
#Task8: Create a mini dataset of your class (at least 5 students):
class_data = {
    "names" :  ["Ubaid", "Ayesha", "haris", "mustaan", "abyan"],
    "Marks" :  [90,80,75,92,70],
    "attendence%" : [95,72,85,90,98]
}
df = pd.DataFrame(class_data)
fig, plots= plt.subplots(1,3, figsize =(18,6))
sns.barplot(x = "names", y = "Marks", data = df, ax = plots[0], color = "purple")
plots[0].set_title("student marks")

sns.lineplot(x = "names", y = "attendence%", data = df, ax = plots[1])
plots[1].set_title("student attendence trend")

correlation_data = df[["Marks","attendence%"]].corr()
sns.heatmap(correlation_data, annot = True,cmap = "coolwarm", ax = plots[2])
plots[2].set_title("(marks vs attendance relationship)")
plt.show()
```
