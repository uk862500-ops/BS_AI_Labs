```python
import random
```

```python
secret = random.randint(1,50)
```

```python
attempt = 0
max_attempt = 7
print("guess a number 1,50: ")
print(f"max_attempt: {max_attempt}")
while attempt < max_attempt:
    try:
        guess = int(input(f" Attempt: {attempt+1}/{max_attempt}, enter a guess"))
    except valueError:
        print("please enter a vaalid number")
        continue
    attempt += 1
    if guess == secret:
        print(f"correct!, you guessed it in {attempt} attempts")
        break
    elif guess < secret:
        print("too low,please try again")
    else:
        print("too high,please try again")
    if attempt == max_attempt and guess != secret:
        print(f"out of attempt! the secret number is {secret}")
```
