```python
import numpy as np
#[Math, Science, English, Computer, Result]
data = np.array([
    [85, 90, 78, 92, 1],
    [42, 48, 35, 50, 0],
    [76, 82, 80, 88, 1],
    [30, 45, 40, 38, 0],[92, 88, 95, 98, 1],
    [55, 60, 58, 62, 0],
    [88, 84, 86, 90, 1],
    [40, 30, 45, 50, 0],
    [65, 70, 68, 75, 1],
    [49, 52, 48, 51, 0]])

X = data[:, :-1]
y = data[:, -1]  
#Calculate average, maximum, and minimum marks.
avg_marks = np.mean(X, axis=0)
max_marks = np.max(X)
min_marks = np.min(X)

#Find number of students who passed and failed.
passed = np.sum(y == 1)
failed = np.sum(y == 0)

#Identify students scoring above 80 and below 50.
above_80 = X[np.any(X > 80, axis=1)]
below_50 = X[np.any(X < 50, axis=1)]

X_min = X.min(axis=0)
X_max = X.max(axis=0)
X_normalized = (X - X_min) / (X_max - X_min)

# Apply simple prediction logic.
predictions = np.where(np.mean(X, axis=1) > 50, 1, 0)

#Calculate accuracy of predictions.
accuracy = np.mean(predictions == y) * 100

#Outputs
print(f"Averge Marksof Subject: {avg_marks}")
print(f"Clas Maximum: {max_marks} | Class Minimum: {min_marks}")
print(f"Studnts Pased: {passed} | Students Failed: {failed}")
print(f"Accuacy of Logc: {accuracy}%")
```
