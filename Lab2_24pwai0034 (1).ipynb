```python
# second part

#Ask for a password (input()).
password = input("enter your password")

if len(password) < 6:
    print("too short!")
else:
    has_digit = False
    for char in password:
        if char.isdigit():
            has_digit = True
            break
            
    if not has_digit:
        print("add at least one number!")
    else:
        has_upper = False
        for char in password:
            if char.isupper():
                has_upper = True
                break
        if not has_upper:
            print("add at least one capital latter!")
        else:
            print("strong password - good choice")
```
